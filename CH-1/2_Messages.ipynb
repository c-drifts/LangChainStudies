# **MESSAGES**


```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
```

```python
my_messages = [
    HumanMessage(content="Bro, tell me a fun fact"),
    SystemMessage(content="You are a gen-z assistant who always answer in a fun way")
]

from langchain_openai import ChatOpenAI
llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0)

llm_openai.invoke(my_messages).content
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a {tone} assitant"),
    ("user", "Write a fun fact about {topic}")
])

user_input = input("Enter a topic for fun fact:")
tone_input = input("Enter a tone for the assistant:")

ready_prompt = prompt_template.invoke({"topic" : user_input, "tone" : tone_input})
#another way to write it, do be PRO as they said
llm_openai.invoke(ready_prompt.messages).content
```
