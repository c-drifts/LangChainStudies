```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI


llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0)
llm_openai.invoke("tell me a joke. generate the output in key-value pair format with the following keys: steup, punchline").content
```

# **Using pydantic models for data validation set** 

```python
from pydantic import BaseModel, Field

#basemodel is a way to build out data validation schema
class LLMSchema(BaseModel):
    setup: str = Field(description="The setup for the joke")
    punchline: str = Field(description="The punchline for the joke")
```

```python
obj = LLMSchema(**{
    "setup"  : "some setup",
    "punchline" : "some puchline"
})

obj
```

```python
llm_structured_ouput = llm_openai.with_structured_output(LLMSchema)
result = llm_structured_ouput.invoke("Tell me a joke")

result.punchline
```

# **Using TypedDict** 

```python
from typing import TypedDict

#basemodel is a way to build out data validation schema
class LLMSchemaTD(TypedDict):
    setup: str 
    punchline: str 
```

```python
obj_td = LLMSchemaTD(**{
    "setup"  : "some setup",
    "punchline" : "some puchline"
})
#here we will obtin not an object, a simple dictiornary
obj_td

#if i want to call it i need to say obj['setup']
#no type match, if there is a typo in the key it will simply return another dictionary
obj_td = LLMSchemaTD(**{
    "ketchup"  : "some setup",
    "punchline" : "some puchline"
})
obj_td
```
